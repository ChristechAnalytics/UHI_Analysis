# Urban Heat Island Effect Analysis in Lagos State
This project analyzes the **Urban Heat Island (UHI)** effect by comparing temperature trends between urban and rural areas of Lagos State in 2024.

**Objective**:
- Identify temperature differences between urban and rural areas.
- Quantify the impact of the UHI effect.

**Data**:
- Simulated daily temperature data for urban and rural areas.
- Timeframe: January 1, 2024 – December 31, 2024.

```python
import pandas as pd

file_path = "C:/Users/Christopher Onyeneke/Desktop/Ch risTech Analytics/Data Science Projects/UHI analysis/urban_heat_island_demo.xlsx"
data = pd.read_excel(file_path)

print(data.head())
```
